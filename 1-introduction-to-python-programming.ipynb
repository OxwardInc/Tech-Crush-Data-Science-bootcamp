# 4. Working with Text (Strings)

A string is a sequence of characters wrapped in `"double quotes"` or `'single quotes'`.
Names, addresses, product descriptions, survey answers and column headings are all strings.

Real data is messy. Names arrive with extra spaces, capital letters are inconsistent, and the same
city is spelled three different ways. Cleaning that text is a large part of a data scientist's day,
and it starts with the tools in this section.


## Joining Strings (Concatenation)
The `+` operator joins two strings into one.


````python
first_name = "Jane"
last_name = "Okoro"

full_name = first_name + " " + last_name
print(full_name)
````

### A Common Error
`+` joins text to text. It cannot join text to a number. Run this and read the error.


````python
student = "Sophia"
age = 45

record = "Student: " + student + " | Age: " + age   # this fails
print(record)
````

### The Fix: Convert the Number
`str()` turns a number into text so `+` can join it.


````python
record = "Student: " + student + " | Age: " + str(age)
print(record)
````

## f-strings: The Cleaner Way
Put an `f` before the opening quote and place variables inside curly brackets.
No `str()` conversion is needed, and the line is much easier to read.


````python
student = "Abena"
age = 30
course = "Data Science"

print(f"Student: {student} | Age: {age} | Course: {course}")

# You can control how many decimal places a number shows
score = 87.456
print(f"Score: {score:.1f}%")
````

## 5. Indexing and Slicing Strings

Every character in a string has a position number, called an index. **Counting starts at 0.**

For the string `"DATA"`:

| Character | D | A | T | A |
|:--|:--|:--|:--|:--|
| Index from the left | 0 | 1 | 2 | 3 |
| Index from the right | -4 | -3 | -2 | -1 |

Negative indexes count backwards from the end, so `-1` is always the last character.


````python
student = "Amaka Bello"

print(student[0])    # first character
print(student[6])    # first letter of the surname
print(student[-1])   # last character
print(len(student))  # how many characters in total
````

### Slicing
Slicing takes a section of a string. The syntax is `text[start:stop]`.

The `start` index is included. The `stop` index is **not** included. This catches everybody once,
so read it twice.


````python
code_value = "DS-2026-047"

print(code_value[0:2])    # DS
print(code_value[3:7])    # 2026
print(code_value[8:])     # from index 8 to the end
print(code_value[:2])     # from the start to index 2
print(code_value[::-1])   # the whole string reversed
````

## 6. String Methods for Cleaning Data

A method is a function that belongs to a value. You call it with a dot: `text.upper()`.

Below is the toolkit you will reach for constantly when cleaning a dataset.

| Method | What it does |
|:--|:--|
| `.strip()` | removes spaces at the start and end |
| `.lower()` | makes everything lowercase |
| `.upper()` | makes everything uppercase |
| `.title()` | capitalises the first letter of each word |
| `.replace(a, b)` | swaps one piece of text for another |
| `.split(sep)` | breaks the text into a list |
| `len(text)` | counts the characters |


````python
messy_name = "   amaka   bello  "

print(f"Original: '{messy_name}'")
print(f"Stripped: '{messy_name.strip()}'")
print(f"Title case: '{messy_name.strip().title()}'")
````

Notice the last line. We called `.strip()` and then `.title()` on the result.
Chaining methods like this is normal Python and saves you from creating a variable at every step.


````python
city = "Lagoss"
print(city.replace("Lagoss", "Lagos"))

record = "Amaka,Bello,Statistics,87"
parts = record.split(",")
print(parts)
print(parts[0])
print(parts[-1])
````

### Why This Matters
Imagine a survey where people typed their city by hand. You receive:
`"lagos"`, `"Lagos "`, `"LAGOS"`, `" lagos"`.

To a human those are one city. To Python they are four different strings, and any count you run will
be wrong. Cleaning the text first is what makes the count correct.


````python
entries = ["  lagos ", "LAGOS", "Lagos", "lagos  "]

cleaned = []
for entry in entries:
    cleaned.append(entry.strip().title())

print("Before:", entries)
print("After: ", cleaned)
````

# 7. Working with Numbers

Python has two number types you will use daily:
- `int` for whole numbers, such as a count of students.
- `float` for numbers with a decimal point, such as a price or an average.


````python
total_students = 45        # int
average_score = 72.8      # float
fee = 15000

print(total_students + 5)
print(average_score * 2)
print(fee / 3)
````

## Arithmetic Operators

| Operator | Meaning | Example | Result |
|:--|:--|:--|:--|
| `+` | add | `7 + 2` | `9` |
| `-` | subtract | `7 - 2` | `5` |
| `*` | multiply | `7 * 2` | `14` |
| `/` | divide, always gives a float | `7 / 2` | `3.5` |
| `//` | floor division, drops the remainder | `7 // 2` | `3` |
| `%` | modulus, gives the remainder | `7 % 2` | `1` |
| `**` | power | `7 ** 2` | `49` |


````python
students = 47
per_group = 5

print(students / per_group)    # 9.4
print(students // per_group)   # 9 full groups
print(students % per_group)    # 2 students left over
print(2 ** 10)
````

`//` and `%` look like small details, but they answer real questions.
How many full buses do I need? How many people are left without a seat?


## Rounding
`round(value, digits)` shortens a decimal. Money and percentages almost always need it.


````python
average = 72.847362

print(round(average))       # nearest whole number
print(round(average, 1))    # one decimal place
print(round(average, 2))    # two decimal places
````

## Decimal Precision: An Odd Result
Run the cell below. The answer is not what you would expect.


````python
print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)
````

Computers store decimals in binary, and some decimals cannot be written exactly in binary,
in the same way that one third cannot be written exactly as a decimal. The result is a very small error.

This is not a Python bug. It happens in almost every language.

**What to do about it:** never test two decimals for exact equality. Round first, or check that the
difference is tiny.


````python
result = 0.1 + 0.2

print(round(result, 2) == 0.3)
print(abs(result - 0.3) < 0.0001)
````

# 9. Booleans and Comparison

A boolean holds one of two values: `True` or `False`. Both start with a capital letter.

Booleans are the answer to a yes or no question, and they are what conditions and filters are built on.


````python
is_registered = True
has_paid = False

print(is_registered)
print(type(is_registered))
````

## Comparison Operators
These compare two values and return `True` or `False`.

| Operator | Meaning |
|:--|:--|
| `==` | equal to |
| `!=` | not equal to |
| `>` | greater than |
| `<` | less than |
| `>=` | greater than or equal to |
| `<=` | less than or equal to |

Note the difference between `=` and `==`. A single `=` **assigns** a value.
A double `==` **compares** two values. Mixing them up is one of the most common beginner errors.


````python
score = 85
pass_mark = 50

print(score > pass_mark)
print(score == 85)
print(score != 85)
print(score <= 70)
````

## Logical Operators
Use these to combine conditions.

| Operator | True when |
|:--|:--|
| `and` | both sides are true |
| `or` | at least one side is true |
| `not` | reverses the result |


````python
score = 85
attendance = 60

print(score >= 50 and attendance >= 75)   # both must be true
print(score >= 50 or attendance >= 75)    # one is enough
print(not score >= 50)
````

# 10. Data Structures

So far each variable has held one value. Real data comes in groups: a class of students,

Python gives you four main containers:

| Structure | Written with | Ordered | Can be changed | Allows duplicates |
|:--|:--|:--|:--|:--|

## 10.1 Lists
A list stores several items in order. It is the container you will use most often.


````python
courses = ["Math", "Physics", "Chemistry", "Biology"]
scores = [85, 72, 91, 60]

print(courses)
print(scores)
print(len(courses))
````

### Accessing Items
Lists use the same indexing rules as strings. Counting starts at 0 and `-1` is the last item.


````python
print(courses[0])      # first item
print(courses[-1])     # last item
print(courses[1:3])    # a slice: index 1 and 2, not 3
````

### A List Can Hold Mixed Types


````python
student = ["Amaka", 21, 87.5, True]

print(student)
print(student[0], "scored", student[2])
````

### Changing a List
Lists are mutable, which means you can change them after they are created.

| Method | What it does |
|:--|:--|
| `.append(x)` | adds `x` to the end |
| `.insert(i, x)` | adds `x` at position `i` |
| `.extend([a, b])` | adds several items at once |
| `.remove(x)` | removes the first `x` it finds |
| `.pop(i)` | removes the item at position `i` and returns it |
| `.sort()` | sorts the list in place |
| `.reverse()` | reverses the list in place |


````python
queue = ["Ada", "Chidi", "Ama"]

queue.append("Kofi")          # add to the end
queue.insert(1, "Bola")       # add at position 1
print(queue)

queue.remove("Ama")           # remove by value
removed = queue.pop(0)        # remove by position
print(queue)
print("Removed:", removed)
````

````python
scores = [85, 72, 91, 60]

scores.sort()
print("Sorted:", scores)

scores.sort(reverse=True)
print("Highest first:", scores)
````

### A Trap Worth Knowing
`.sort()` changes the original list and returns nothing. `sorted()` leaves the original alone and
gives you a new list. Confusing the two is a very common bug.


````python
original = [3, 1, 2]

new_list = sorted(original)
print("original:", original)
print("new_list:", new_list)

result = original.sort()
print("result of .sort():", result)   # None
print("original is now:", original)
````

### Lists Inside Lists
A list of lists is how you store a small table. Each inner list is one row.


````python
records = [
    ["Amaka", 87],
    ["Chidi", 62],
    ["Bola", 74],
]

print(records[0])       # the first row
print(records[0][0])    # name in the first row
print(records[1][1])    # score in the second row
````

## 10.2 Dictionaries
A dictionary stores pairs: a **key** and a **value**. Instead of remembering that position 2 holds
the score, you ask for the key `"score"` by name.

This is how one record is usually stored, and it is very close to how a row of a real dataset looks.


````python
student = {
    "name": "Amaka",
    "age": 21,
    "course": "Data Science",
    "score": 87.5
}

print(student["name"])
print(student["score"])
````

### Adding and Updating


````python
student["level"] = "300 Level"   # add a new pair
student["score"] = 91.0          # update an existing one

print(student)
````

### Safe Access with .get()
Asking for a key that does not exist raises an error. `.get()` returns `None` instead,
or a default value that you choose.


````python
print(student.get("course"))
print(student.get("phone"))                  # None, no error
print(student.get("phone", "not provided"))  # your own default
````

````python
print(student["phone"])   # this raises a KeyError
````

### Useful Dictionary Methods


````python
print(student.keys())
print(student.values())
print(student.items())

# Removing a pair
student.pop("level")
print(student)
````

## 10.3 Tuples
A tuple looks like a list but uses round brackets, and it cannot be changed after it is created.
That is the whole point: use a tuple when the values should stay fixed.

Good uses: a screen resolution, a pair of coordinates, a set of fixed categories,

````python
group_a = {"Ada", "Chidi", "Bola"}
group_b = {"Bola", "Kofi", "Ada"}

print(group_a | group_b)   # union: everyone
print(group_a & group_b)   # intersection: in both
print(group_a - group_b)   # difference: only in A
````

# 11. Making Decisions: if, elif, else

A condition lets your code take different paths.

```python
if condition:
    # runs when the condition is True
elif another_condition:
    # runs when the first was False and this one is True
else:
    # runs when nothing above was True
```

Two rules that Python enforces strictly:
1. The line ends with a colon `:`.
2. The block under it is **indented**, normally by four spaces.


````python
score = 85

if score >= 70:
    print("Distinction")
elif score >= 50:
    print("Pass")
else:
    print("Fail")
````

### Order Matters
Python checks the conditions from top to bottom and stops at the first true one.
If you put `score >= 50` first, a score of 85 would print "Pass" and never reach "Distinction".


````python
score = 85

# The wrong order
if score >= 50:
    print("Pass")
elif score >= 70:
    print("Distinction")   # this can never be reached
````

### Watch Your Indentation
Indentation is not decoration in Python. It defines which lines belong to the block.
Run this to see the error.


````python
score = 85

if score >= 70:
print("Distinction")
````

# 12. Loops

A loop repeats work so you do not have to write the same line fifty times.
This is the point where code starts to save you real effort.

## The for Loop
A `for` loop walks through every item in a collection.


````python
scores = [45, 85, 60, 92]

for score in scores:
    if score >= 50:
        print(f"{score} - Pass")
    else:
        print(f"{score} - Fail")
````

### range()
`range(n)` produces the numbers from 0 up to but not including `n`.
Use it when you want to repeat something a fixed number of times.


````python
for i in range(5):
    print(f"Processing record {i + 1}")

print("---")

for i in range(2, 10, 2):   # start, stop, step
    print(i)
````

### Looping Through a Dictionary


````python
student = {"name": "Amaka", "age": 21, "score": 87.5}

for key in student:
    print(key, "->", student[key])

print("---")

for key, value in student.items():
    print(f"{key}: {value}")
````

### Building a Result Inside a Loop
This pattern comes up constantly: start with an empty list or a counter,
then add to it on each pass.


````python
scores = [45, 85, 60, 92, 38]

passed = []
total = 0

for score in scores:
    total = total + score
    if score >= 50:
        passed.append(score)

print("Passed:", passed)
print("How many passed:", len(passed))
print("Average:", round(total / len(scores), 2))
````

## The while Loop
A `while` loop repeats for as long as a condition stays true. Use it when you do not know in
advance how many times you need to repeat.

Make sure something inside the loop moves the condition towards `False`.
If it never does, the loop runs forever and you have to stop the cell by hand.


````python
papers_left = 3

while papers_left > 0:
    print(f"Marking paper {papers_left}")
    papers_left = papers_left - 1

print("Marking complete")
````

### break and continue
- `break` leaves the loop immediately.
- `continue` skips the rest of this pass and moves to the next item.


````python
for score in [45, 85, 0, 92, 60]:
    if score == 0:
        continue          # skip missing entries
    if score > 90:
        print("Top score found:", score)
        break
    print("Checked:", score)
````

# 14. A First Look at List Comprehension

A list comprehension builds a new list in one line. It does exactly what a `for` loop does,
written more compactly.

```python
new_list = [what_to_keep for item in collection if condition]
```

We cover this properly in Week 2. See it once now so the syntax is familiar when it returns.


````python
scores = [45, 85, 60, 92, 38]

# The long way
passed = []
for score in scores:
    if score >= 50:
        passed.append(score)
print("Loop:         ", passed)

# The same thing as a comprehension
passed = [score for score in scores if score >= 50]
print("Comprehension:", passed)
````

````python
# Changing every item
doubled = [score * 2 for score in scores]
print(doubled)

# Labelling every item
labels = ["Pass" if score >= 50 else "Fail" for score in scores]
print(labels)
````

# 15. Mini Project: A Small Sales Report

Everything in this notebook comes together here. Read the code, run it,
then change the numbers and run it again.

The dataset below is a list of dictionaries, which is the shape real data usually arrives in.


````python
sales = [
    {"product": "Solar Lamp",  "category": "Electronics", "units": 37, "price": 12500},
    {"product": "Rice (5kg)",  "category": "Food",        "units": 120, "price": 8500},
    {"product": "Ankara Fabric","category": "Fashion",    "units": 54, "price": 6000},
    {"product": "Power Bank",  "category": "Electronics", "units": 22, "price": 15000},
    {"product": "Garri (2kg)", "category": "Food",        "units": 95, "price": 2200},
]

print("Number of products:", len(sales))
print("First record:", sales[0])
````

### Step 1: Revenue per product


````python
for item in sales:
    revenue = item["units"] * item["price"]
    item["revenue"] = revenue          # store it back in the record
    print(f'{item["product"]:<16} {revenue:>10,}')
````

### Step 2: Totals and the best seller


````python
total_revenue = 0
best_product = ""
best_revenue = 0

for item in sales:
    total_revenue = total_revenue + item["revenue"]
    if item["revenue"] > best_revenue:
        best_revenue = item["revenue"]
        best_product = item["product"]

print(f"Total revenue: {total_revenue:,}")
print(f"Best seller:   {best_product} ({best_revenue:,})")
print(f"Average per product: {round(total_revenue / len(sales), 2):,}")
````

### Step 3: Revenue by category
A dictionary is the right tool here, because each category name is a key.


````python
by_category = {}

for item in sales:
    category = item["category"]
    if category in by_category:
        by_category[category] = by_category[category] + item["revenue"]
    else:
        by_category[category] = item["revenue"]

for category, revenue in by_category.items():
    share = round(revenue / total_revenue * 100, 1)
    print(f"{category:<14} {revenue:>10,}  ({share}%)")
````

### Your turn: finish the report
1. Print the name of every product that sold more than 50 units.
2. Find the product with the **lowest** revenue.
3. Print how many unique categories there are.


````python
# START HERE
for item in sales:
    if item["units"] > 50:
        print(item["product"])

print(min(sales, key=lambda item: item["units"] * item["price"]))

print(set(item["category"] for item in sales))
````
