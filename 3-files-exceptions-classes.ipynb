# 1. Opening a File and File Modes

When your notebook closes, every variable disappears. A **file** keeps data on the disk, so it is still
there tomorrow.

```python
with open("filename.txt", "mode") as f:
    # work with the file here
```

| Mode | Meaning | If the file exists | If the file does not exist |
|:--|:--|:--|:--|
| `"r"` | read (the default) | reads it | error |
| `"w"` | write | **erases it first** | creates it |
| `"a"` | append | adds to the end | creates it |

**Why `with`?** A file must be closed after use, or your work may not be saved.
`with` closes it for you automatically when the indented block ends.

# 2. Writing and Appending


````python
with open("report.txt", "w") as f:
    f.write("KADO STORES SALES REPORT\n")
    for item in sales:
        f.write(f"{item['product']}: {revenue(item)}\n")

print("Report saved")
````

Two rules for `write()`:
1. It does **not** start a new line for you. Add `\n` at the end of each line.
2. It only accepts **text**. The f-string turns numbers into text for you.

Now **append** a line. `"a"` adds to the end and keeps everything already in the file.


````python
with open("report.txt", "a") as f:
    f.write("Checked by: Manager\n")

with open("report.txt", "r") as f:
    print(f.read())
````

# 4. From a CSV File to Records

Real data usually comes as a **CSV** file: comma separated values. The first line is the header.
Let us create one, then read it back.


````python
with open("sales.csv", "w") as f:
    f.write("product,category,units,price\n")
    for item in sales:
        f.write(f"{item['product']},{item['category']},{item['units']},{item['price']}\n")

with open("sales.csv", "r") as f:
    print(f.read())
````

````python
records = []

with open("sales.csv", "r") as f:
    next(f)                                    # skip the header line
    for line in f:
        product, category, units, price = line.strip().split(",")
        records.append({"product": product, "category": category,
                        "units": int(units), "price": int(price)})

print(records[0])
````

**Everything read from a file is text.** That is why we wrap units and price in `int()`.
Forget that, and `"37" * 2` gives you `"3737"` instead of `74`.

# 5. When Code Meets Bad Data

Real files contain mistakes. Run this cell. It fails on purpose.


````python
# This cell fails on purpose
units = ["37", "120", "twenty two", "45"]

for value in units:
    print(int(value))
````

The loop printed 37 and 120, then stopped. One bad value stopped everything, and 45 was never reached.

An error that happens while code is running is called an **exception**.

| Exception | Usual cause |
|:--|:--|
| `ValueError` | right type, wrong value: `int("twenty two")` |
| `TypeError` | wrong type: `"5" + 5` |
| `KeyError` | missing dictionary key |
| `IndexError` | list position that does not exist |
| `ZeroDivisionError` | dividing by zero |
| `FileNotFoundError` | the file does not exist |
| `AttributeError` | asking an object for something it does not have (you will meet this in Part 3) |

**Tip:** always read the **last line** of an error first. It names the problem.

# 6. try and except

```python
try:
    # the line that might fail
except SomeError:
    # what to do instead
```


````python
clean = []

for value in units:
    try:
        clean.append(int(value))
    except ValueError:
        print(f"Skipping bad entry: {value}")

print(clean)
````

This time the loop skipped the bad value and carried on to 45.

**Always name the error you expect.** A bare `except:` catches every error, including typing mistakes
in your own code, so real bugs get hidden.

# 7. Handling Different Errors

One `try` can have several `except` blocks. Python runs the first one that matches.


````python
def average_price(total, count):
    try:
        return total / int(count)
    except ValueError:
        print("count is not a number")
    except ZeroDivisionError:
        print("count cannot be zero")

print(average_price(36000, "3"))
print(average_price(36000, "three"))
print(average_price(36000, "0"))
````

When an error is caught, the function never reaches `return`, so it gives back `None`.

---
# 8. else and finally

| Block | When it runs |
|:--|:--|
| `try` | always attempted |
| `except` | only if an error happened |
| `else` | only if **no** error happened |
| `finally` | always, at the very end |


````python
def load_report(filename):
    try:
        with open(filename, "r") as f:
            text = f.read()
    except FileNotFoundError:
        print(f"{filename} is missing")
    else:
        print(f"{filename} loaded: {len(text)} characters")
    finally:
        print("Finished checking", filename)

load_report("report.txt")
print("---")
load_report("missing.txt")
````

# 9. Why Do We Need Classes?

So far, a product has been a **dictionary**, and the things we do with it are separate **functions**.


````python
lamp = {"name": "Solar Lamp", "units": 37, "price": 12500}

def product_revenue(product):
    return product["units"] * product["price"]

print(product_revenue(lamp))
````

This works. But look at the problems it can cause.


````python
# This cell fails on purpose
rice = {"name": "Rice (5kg)", "unit": 120, "price": 8500}   # typo: "unit" instead of "units"

print(product_revenue(rice))
````

Nothing stopped us from creating a product with a misspelled key. We only found out later,
when a calculation broke. Three more problems:

1. **Nothing connects the data to the actions.** `product_revenue()` is just a function floating
   around. You must remember it exists and that it works on products.
2. **Nothing guarantees the shape.** One product might have a `"stock"` key, another might not.
3. **It gets messy as it grows.** Ten product functions means ten loose functions to keep track of.

A **class** solves all three. It is a way to say: "every product **has** a name, units and price,
and every product **can** calculate its revenue." The data and the actions live together, in one place.


---
# 10. The Big Idea: a Blueprint and the Things Built From It

Think of a **cookie cutter**.

- The **cookie cutter** is the **class**. It describes the shape. You cannot eat a cookie cutter.
- Each **cookie** is an **object**. You can make as many cookies as you want from one cutter.
- Every cookie has the same shape, but each cookie can have its **own** topping.

Or think of a **registration form** at school.

- The **blank form** is the **class**. It says every student has a name, an age and a class.
- Each **filled form** is an **object**. Amaka's form and Tolu's form follow the same layout,
  but hold different details.

| Word | Meaning | Cookie example | School example |
|:--|:--|:--|:--|
| **class** | the blueprint | the cookie cutter | the blank form |
| **object** | one thing made from it | one cookie | Amaka's filled form |
| **attribute** | data an object holds | its topping | Amaka's age |
| **method** | an action an object can do | (cookies cannot do much) | "calculate my average" |

You have been using objects since Week 1 without knowing it.


````python
word = "lagos"
numbers = [3, 1, 2]

print(type(word))       # word is an object of the class str
print(type(numbers))    # numbers is an object of the class list

print(word.upper())     # upper() is a METHOD of the str class
numbers.append(4)       # append() is a METHOD of the list class
print(numbers)
````

`"lagos"` is one object made from the `str` blueprint. `[3, 1, 2]` is one object made from the
`list` blueprint. The dot in `word.upper()` means "ask this object to do its `upper` action".

Today you learn to **write your own blueprint**.

# 11. The Smallest Class Possible

```python
class Product:
    pass
```

- `class` tells Python you are creating a blueprint.
- `Product` is its name. Class names start with a **capital letter**. This is a convention that
  helps you tell classes apart from functions and variables.
- `pass` means "nothing here yet".


````python
class Product:
    pass

lamp = Product()     # make one object from the blueprint
rice = Product()     # make another

print(type(lamp))
print(lamp == rice)  # two different objects
````

`Product()` with brackets **creates an object**. We now have two products. They are separate
things, even though they came from the same blueprint.

But they are empty. We can stick data onto each one with a dot:


````python
lamp.name = "Solar Lamp"
lamp.units = 37
lamp.price = 12500

rice.name = "Rice (5kg)"
rice.units = 120
rice.price = 8500

print(lamp.name, lamp.units)
print(rice.name, rice.units)
````

Each object now holds its **own** data. `lamp.name` and `rice.name` are different.
These values attached with a dot are called **attributes**.

But this is painful. Three lines for every product, and nothing stops you from forgetting one.
We want the data to be set up **automatically** every time a product is created. That is what the next
section does.

# 12. `__init__`: Setting Up a New Object

`__init__` is a special method that Python runs **automatically** the moment an object is created.
Its job is to set up the object's data. Think of it as **filling in the form**.

It is written with **two underscores on each side**: `__init__`. People often type one underscore by
mistake, and then it does nothing.


````python
class Product:
    def __init__(self, name, units, price):
        self.name = name
        self.units = units
        self.price = price

lamp = Product("Solar Lamp", 37, 12500)
rice = Product("Rice (5kg)", 120, 8500)

print(lamp.name, lamp.units, lamp.price)
print(rice.name, rice.units, rice.price)
````

One line now creates a complete product. Here is exactly what happens when Python runs
`lamp = Product("Solar Lamp", 37, 12500)`:

| Step | What Python does |
|:--|:--|
| 1 | Creates a new, empty Product object |
| 2 | Calls `__init__` automatically, passing the new object in as `self` |
| 3 | `name` becomes `"Solar Lamp"`, `units` becomes `37`, `price` becomes `12500` |
| 4 | `self.name = name` stores `"Solar Lamp"` **on the object** |
| 5 | Same for `self.units` and `self.price` |
| 6 | Hands the finished object back, and it is stored in `lamp` |

### Why `self.name = name`, and not just `name`?

- `name` on its own is a normal variable inside `__init__`. Like every local variable from Day 1,
  it **disappears** when `__init__` finishes.
- `self.name` is stored **on the object**, so the object remembers it forever.

`self.name = name` means: "take the value that came in, and save it on this object."

# 13. What Is `self`?

`self` is the word most students find confusing, so let us slow down.

**`self` means "this particular object".**

Think about how people talk. When Amaka says "**my** name is Amaka" and Tolu says "**my** name is Tolu",
they use the same word, "my", but it points to a different person each time. `self` is Python's "my".

Inside the class, `self.name` means "**my** name": the name of whichever object is being used right now.

- When Python is working with `lamp`, `self` **is** `lamp`.
- When Python is working with `rice`, `self` **is** `rice`.

This is how one blueprint can serve many objects. The code is written once, and `self` decides
whose data it touches.


### Proof: `self` really is the object

Python gives every object a unique ID number with `id()`. Watch what happens.


````python
class Product:
    def __init__(self, name, units, price):
        print("Inside __init__, self has ID:", id(self))
        self.name = name
        self.units = units
        self.price = price

lamp = Product("Solar Lamp", 37, 12500)
print("Outside, lamp has ID:       ", id(lamp))
````

The two numbers are the **same**. While `__init__` was running, `self` and `lamp` were the
same object. `self` is simply the name the class uses for "the object I am working on right now".

### You never pass `self` yourself
Look at the call: `Product("Solar Lamp", 37, 12500)`. That is **three** arguments, but `__init__`
has **four** parameters: `self, name, units, price`.

Python fills in `self` for you, automatically. You only pass the rest.

# 15. Methods: the Actions an Object Can Do

A **method** is a function written **inside** a class. Every method takes `self` as its **first**
parameter, so it can reach the object's own data.


````python
class Product:
    def __init__(self, name, units, price):
        self.name = name
        self.units = units
        self.price = price

    def revenue(self):
        return self.units * self.price

lamp = Product("Solar Lamp", 37, 12500)
rice = Product("Rice (5kg)", 120, 8500)

print(lamp.revenue())
print(rice.revenue())
````

Compare this with Section 9:

| Before: dictionary + loose function | Now: class with a method |
|:--|:--|
| `product_revenue(lamp)` | `lamp.revenue()` |
| You pass the product **into** the function | You ask the product **to do it** |
| The function lives somewhere else | The method lives **inside** the class |

`lamp.revenue()` reads like a sentence: "lamp, give me your revenue."


### What Python does behind the scenes

When you write `lamp.revenue()`, Python secretly turns it into `Product.revenue(lamp)`.
That is where `self` comes from: the object **before the dot** becomes `self`.


````python
print(lamp.revenue())            # the normal way
print(Product.revenue(lamp))     # what Python actually does
````

Both lines give the same answer. You will always write the first one, but now you know why
`self` is there, and why you never pass it yourself.

### Brackets matter


````python
print(lamp.revenue)      # the method itself: no brackets, so it does not run
print(lamp.revenue())    # the result: brackets make it run
````

Same rule as functions in Day 1: no brackets means "the thing itself", brackets mean "run it".

# 16. Methods That Take Extra Inputs

A method can take inputs, just like a function. `self` always comes first, then your own parameters.


````python
class Product:
    def __init__(self, name, units, price):
        self.name = name
        self.units = units
        self.price = price

    def revenue(self):
        return self.units * self.price

    def price_after_discount(self, percent):
        return self.price - self.price * percent / 100

lamp = Product("Solar Lamp", 37, 12500)

print(lamp.price_after_discount(10))
print(lamp.price_after_discount(20))
````

`price_after_discount(self, percent)` has two parameters, but we only pass **one** argument,
`10`. Python fills in `self` with `lamp`, and `10` goes into `percent`.

# 17. Methods That Change the Object

So far our methods only **calculated** something. A method can also **change** the object's own data.
This is where classes become really useful: the object **remembers** what happened to it.

Let us give our product a `stock` count, and methods to sell and restock.


````python
class Product:
    def __init__(self, name, stock, price):
        self.name = name
        self.stock = stock
        self.price = price

    def sell(self, qty):
        if qty > self.stock:
            print(f"Cannot sell {qty}. Only {self.stock} {self.name} left.")
            return
        self.stock = self.stock - qty
        print(f"Sold {qty} {self.name}. {self.stock} left.")

    def restock(self, qty):
        self.stock = self.stock + qty
        print(f"Added {qty} {self.name}. {self.stock} in stock.")

lamp = Product("Solar Lamp", 10, 12500)
print("Starting stock:", lamp.stock)
````

### Predict first
Before you run the next cell, work out the stock after each line. Write your four numbers down.


````python
lamp.sell(4)
lamp.sell(9)
lamp.restock(20)
lamp.sell(9)
print("Final stock:", lamp.stock)
````

<details>
  <summary>Click for the explanation</summary>

| Line | What happens | Stock after |
|:--|:--|:--|
| start | | 10 |
| `sell(4)` | 4 is not more than 10, so take 4 away | 6 |
| `sell(9)` | 9 is more than 6, so print a warning and change nothing | 6 |
| `restock(20)` | add 20 | 26 |
| `sell(9)` | take 9 away | 17 |

The lamp **remembered** its stock between every call. A normal function forgets everything when it
ends. An object keeps its attributes for as long as it exists.
</details>

### Each object keeps its own count


````python
lamp = Product("Solar Lamp", 10, 12500)
rice = Product("Rice (5kg)", 50, 8500)

lamp.sell(3)
print("Lamp stock:", lamp.stock)
print("Rice stock:", rice.stock)
````

# 20. Objects Inside Objects: a Shopping Cart

An attribute can hold **anything**: a number, a string, a list, or even other objects.
A shopping cart is a good example. The cart is one object, and it holds a list of Product objects.


````python
class Cart:
    def __init__(self, customer):
        self.customer = customer
        self.items = []                  # starts empty

    def add(self, product, qty):
        self.items.append((product, qty))
        print(f"Added {qty} x {product.name}")

    def total(self):
        amount = 0
        for product, qty in self.items:
            amount = amount + product.price * qty
        return amount

cart = Cart("Amaka")
cart.add(products[0], 2)     # 2 Solar Lamps
cart.add(products[1], 3)     # 3 bags of rice

print(f"{cart.customer} owes {cart.total():,} naira")
````

Things to notice:
- `self.items = []` does not come from an argument. `__init__` can also set **starting values**.
- Each item in the list is a **tuple** `(product, qty)`, and `total()` unpacks it, just like Week 1.
- `product.price` reaches into a Product object that lives inside the Cart object.

# 21. Class Mistakes Everyone Makes

Each cell below fails on purpose. Run each one, read the **last line** of the error, then read the fix.

### Mistake 1: forgetting `self` in a method


````python
# This cell fails on purpose
class Box:
    def __init__(self, size):
        self.size = size

    def describe():                  # self is missing
        return "A box"

b = Box(3)
b.describe()
````

**Error:** `describe() takes 0 positional arguments but 1 was given`

Python always passes the object in as the first argument. With no `self` to receive it, Python
complains. **Fix:** `def describe(self):`

### Mistake 2: forgetting `self.` when saving data


````python
# This cell fails on purpose
class Box:
    def __init__(self, size):
        size = size                  # should be self.size = size

b = Box(3)
print(b.size)
````

**Error:** `'Box' object has no attribute 'size'`

`size = size` only made a local variable that disappeared when `__init__` ended. Nothing was saved
on the object. **Fix:** `self.size = size`

### Mistake 3: one underscore instead of two


````python
# This cell fails on purpose
class Box:
    def _init_(self, size):          # one underscore on each side
        self.size = size

b = Box(3)
````

**Error:** `Box() takes no arguments`

Python only runs a method called `__init__` (two underscores each side) automatically. `_init_` is just
an ordinary method that never gets called. **Fix:** `def __init__(self, size):`

### Mistake 4: the wrong number of arguments


````python
# This cell fails on purpose
class Box:
    def __init__(self, size, colour):
        self.size = size
        self.colour = colour

b = Box(3)
````

**Error:** `__init__() missing 1 required positional argument: 'colour'`

`__init__` asks for `size` and `colour`, but we only gave `size`. **Fix:** `Box(3, "red")`, or give
`colour` a default value: `def __init__(self, size, colour="brown"):`

### Mistake 5: using the class instead of an object


````python
# This cell fails on purpose
class Box:
    def __init__(self, size):
        self.size = size

print(Box.size)
````

**Error:** `type object 'Box' has no attribute 'size'`

The blueprint does not have a size. Only a real box does. You cannot measure a cookie cutter's topping.
**Fix:** create an object first: `b = Box(3)`, then `print(b.size)`.

# 22. Dictionary or Class?

Both can hold a record. Here is how to choose.

| | Dictionary | Class |
|:--|:--|:--|
| Create one | `{"name": "Solar Lamp", "price": 12500}` | `Product("Solar Lamp", 12500)` |
| Read a value | `item["price"]` | `item.price` |
| Has its own actions | No, you write separate functions | Yes, methods live inside |
| Guarantees the same shape | No, any key can be missing | Yes, `__init__` sets every attribute |
| Best for | quick, simple data, data from files | things with **behaviour**: selling, restocking, calculating |

In data science you will meet both. Pandas reads a CSV into a **DataFrame**, which is an object made
from the `DataFrame` class. When you write `df.head()` next week, you are calling a method on an object,
exactly like `lamp.revenue()`.

# 23. Mini Project: Load a Messy File into Objects

This joins all three parts of today: **files**, **exceptions** and **classes**.

Run this cell to create a messy stock file. It has a word where a number should be, and a missing value.


````python
with open("stock.csv", "w") as f:
    f.write("product,stock,price\n")
    f.write("Solar Lamp,10,12500\n")
    f.write("Rice (5kg),50,8500\n")
    f.write("Ankara Fabric,twelve,6000\n")
    f.write("Power Bank,8,15000\n")
    f.write("Garri (2kg),,2200\n")
    f.write("Palm Oil (1L),30,3500\n")

with open("stock.csv", "r") as f:
    print(f.read())
````

### Let us build it together, one step at a time

**Step 1:** the class. Same `Product` as Section 18, with `value()` and `__str__`.


````python
class Product:
    def __init__(self, name, stock, price):
        self.name = name
        self.stock = stock
        self.price = price

    def value(self):
        return self.stock * self.price

    def __str__(self):
        return f"{self.name} | stock: {self.stock} | price: {self.price:,}"
````

**Step 2:** a function that reads the file and turns each good row into a Product object.
- The **outer** `try` handles a missing file.
- The **inner** `try` handles one bad row, so the loop skips it and carries on.


````python
def load_products(filename):
    """Read a stock CSV and return a list of Product objects."""
    products = []
    try:
        with open(filename, "r") as f:
            next(f)                                       # skip the header
            for line in f:
                name, stock, price = line.strip().split(",")
                try:
                    products.append(Product(name, int(stock), int(price)))
                except ValueError:
                    print(f"Skipping bad row: {line.strip()}")
    except FileNotFoundError:
        print(f"File not found: {filename}")
    return products

products = load_products("stock.csv")
print(f"\nLoaded {len(products)} products:")
for p in products:
    print(p)
````

Four products loaded, two bad rows skipped, and nothing crashed.

**Step 3:** write a report back to a file.


````python
total = 0
with open("stock_report.txt", "w") as f:
    f.write("KADO STORES STOCK VALUE REPORT\n")
    for p in products:
        f.write(f"{p.name}: {p.value():,}\n")
        total = total + p.value()
    f.write(f"TOTAL: {total:,}\n")

with open("stock_report.txt", "r") as f:
    print(f.read())
````

# Wrap Up

## Classes in Six Lines
1. A **class** is a blueprint. An **object** is one thing made from it.
2. `__init__` runs automatically when an object is created, and sets up its data.
3. `self` means "this particular object". Python passes it in for you.
4. `self.name = name` saves a value **on the object**, so it is remembered.
5. A **method** is a function inside a class. It always takes `self` first.
6. `lamp.revenue()` is really `Product.revenue(lamp)`: the object before the dot becomes `self`.

## Files and Exceptions in Four Lines
1. `with open()` closes the file for you. `"w"` erases, `"a"` adds, `"r"` reads.
2. Everything read from a file is text, so convert numbers with `int()` or `float()`.
3. `try` and `except` let bad data fail quietly instead of stopping the program.
4. Name the error you expect. Do not use a bare `except:`.
