# 2. Why Functions?

Look at the code below. It calculates revenue for three products.

# 3. Defining and Calling a Function

```python
def function_name(parameters):
    # the body: the work the function does
    return result
```

| Part | Meaning |
|:--|:--|
| `def` | tells Python you are defining a function |
| `function_name` | the name you choose, in snake_case |
| `(parameters)` | the inputs the function needs |
| `:` | ends the first line |
| the indented body | the work the function does |
| `return` | sends a result back to whoever called the function |


````python
def revenue(units, price):
    total = units * price
    return total

print("Function defined")
````

Running that cell printed "Function defined", but no revenue was calculated.
**Defining** a function only teaches Python the recipe. Nothing happens until you **call** it by name
with brackets.


````python
print(revenue(37, 12500))
print(revenue(120, 8500))

result = revenue(22, 15000)
print(f"Power Bank revenue: {result:,}")
````

and in the call.

# 4. Parameters and Arguments

These two words are often mixed up.

- A **parameter** is the name in the `def` line. It is a placeholder.
- An **argument** is the actual value you pass in when you call the function.

```python
def revenue(units, price):    # units and price are parameters
    return units * price

revenue(37, 12500)            # 37 and 12500 are arguments
```

Arguments fill the parameters **in order**, from left to right. These are called positional arguments.


````python
def describe(product, units):
    return f"{product} sold {units} units"

print(describe("Solar Lamp", 37))
print(describe(37, "Solar Lamp"))   # wrong order: Python does not know, it just follows the order
````

# 5. Default Values and Keyword Arguments

A **default value** makes a parameter optional. If the caller leaves it out, the default is used.


````python
def label(units, threshold=50):
    if units > threshold:
        return "BEST SELLER"
    return "REGULAR"

print(label(37))       # threshold is 50, the default
print(label(37, 30))   # threshold is 30
````

### Keyword Arguments
You can name the parameter when you call the function. This is a **keyword argument**.
It makes the call easier to read, and the order no longer matters.


````python
print(label(37, threshold=30))
print(label(threshold=30, units=37))   # same result, different order
````

### One Rule
Parameters with default values must come **after** parameters without them.
This cell fails on purpose.


````python
# This cell fails on purpose
def bad_label(threshold=50, units):
    return units > threshold
````

# 6. return versus print

This is the most important idea in this notebook, and the most common source of beginner bugs.

- `print()` **shows** a value on the screen for a human to read. The value is then gone.
- `return` **hands the value back** to the code that called the function, so it can be stored and used.


````python
def revenue_print(units, price):
    print(units * price)       # shows it, gives nothing back

def revenue_return(units, price):
    return units * price       # hands it back

a = revenue_print(37, 12500)
b = revenue_return(37, 12500)

print("a is:", a)
print("b is:", b)
````

Both functions displayed 462500 at some point, but only `b` holds the value.
A function without `return` gives back `None`, which means "nothing".

Now watch what happens when you try to use the result.


````python
# This cell fails on purpose
total = revenue_print(37, 12500) + revenue_print(120, 8500)
````

````python
total = revenue_return(37, 12500) + revenue_return(120, 8500)
print(f"Combined: {total:,}")
````

### return Ends the Function
As soon as Python reaches `return`, the function stops. Any line after it never runs.


````python
def check_stock(stock):
    if stock == 0:
        return "OUT OF STOCK"
    return "AVAILABLE"
    print("This line never runs")


print(check_stock(0))
print(check_stock(5))
````

# 7. Returning More Than One Value

A function can return several values separated by commas. Python packs them into a **tuple**.
You then unpack the tuple into separate variables in one line, exactly as you did with tuples last week.


````python
def summary(sales):
    total_units = 0
    total_rev = 0
    for item in sales:
        total_units += item["units"]
        total_rev += item["units"] * item["price"]
    return total_units, total_rev

result = summary(sales)
print("As a tuple:", result)

units, rev = summary(sales)
print(f"Units: {units} | Revenue: {rev:,}")
````

# 9. Scope: Where a Variable Lives

A variable created **inside** a function is **local**. It exists only while the function runs,
and disappears when the function ends.

A variable created **outside** any function is **global**. Functions can read it.


````python
store = "Kado Stores"          # global

def report(units, price):
    report_total = units * price      # local
    return f"{store}: {report_total:,}"

print(report(37, 12500))
````

The function could read `store`, because it is global. Now try to read `report_total` from outside.
This cell fails on purpose.


````python
# This cell fails on purpose
print(report_total)
````

This is a feature, not a problem. Each function keeps its own variables to itself,
so two functions can both use a variable called `total` without breaking each other.

**The rule to follow:** pass values **in** with arguments and send them **out** with `return`.

# 10. Building a Program from Small Functions

Good programs are built from small functions that each do **one** job.
Functions can call other functions.


````python
def revenue(item):
    return item["units"] * item["price"]

def label(item, threshold=50):
    if item["units"] > threshold:
        return "BEST SELLER"
    return "REGULAR"

def print_report(sales):
    print(f"{'PRODUCT':<14} {'REVENUE':>10}  LABEL")
    for item in sales:
        print(f'{item["product"]:<14} {revenue(item):>10,}  {label(item)}')

print_report(sales)
````

Notice how `print_report()` reads almost like English. Each helper function has a clear name,
so you can understand the report without reading every calculation.

If the revenue rule changes, you edit `revenue()` and the report updates automatically.

# 11. Lambda Functions

A **lambda** is a small function written in one line, without a name.

```python
lambda inputs: result
```

- No `def`, no function name, no `return`.
- It holds **one expression**, and the result of that expression is returned automatically.


````python
# The normal way
def add_vat(price):
    return price * 1.075

# The same thing as a lambda
add_vat_lambda = lambda price: price * 1.075

print(add_vat(10000))
print(add_vat_lambda(10000))
````

A lambda can take more than one input, and it can use a one line `if ... else`.


````python
revenue_l = lambda units, price: units * price
print(revenue_l(37, 12500))

label_l = lambda units: "BEST SELLER" if units > 50 else "REGULAR"
print(label_l(120))
print(label_l(22))
````

Storing a lambda in a variable, as above, is useful for learning. In real code, a named function
is usually written with `def`. Lambda earns its place when you pass it **straight into** another
function, as the next sections show.

# 12. Lambda with sorted()

Last week you used `sorted()` on a list of numbers. Try it on our list of dictionaries.
This cell fails on purpose.


````python
# This cell fails on purpose
sorted(sales)
````

Python does not know how to compare two dictionaries. Is one product "bigger" because of units?
Price? Name?

You tell it with the `key` argument. `key` takes a function that picks the value to compare.
A lambda is perfect for this.


````python
by_units = sorted(sales, key=lambda item: item["units"], reverse=True)

for item in by_units:
    print(f'{item["product"]:<14} {item["units"]}')
````

Read `key=lambda item: item["units"]` as: **for each item, sort by its units.**

`reverse=True` puts the highest first.

You can sort by anything you can calculate, not only by a stored value.


````python
by_revenue = sorted(sales, key=lambda item: item["units"] * item["price"], reverse=True)

for item in by_revenue:
    print(f'{item["product"]:<14} {item["units"] * item["price"]:>10,}')
````

# 13. Lambda with filter() and map()

- `filter(function, data)` keeps only the items where the function returns `True`.
- `map(function, data)` applies the function to **every** item.

Both give back a special object, so wrap them in `list()` to see the result.


````python
# filter: keep the strong sellers
strong = list(filter(lambda item: item["units"] > 50, sales))

for item in strong:
    print(item["product"], item["units"])
````

````python
# map: pull one value out of every record
names = list(map(lambda item: item["product"], sales))
print(names)

# map: change every value
prices_with_vat = list(map(lambda item: item["price"] * 1.075, sales))
print(prices_with_vat)
````

# 14. def or lambda?

| | `def` | `lambda` |
|:--|:--|:--|
| Has a name | Yes | No, unless you assign it |
| Length | As many lines as you need | One expression only |
| Can hold `for`, `print`, several steps | Yes | No |
| Docstring | Yes | No |
| Best for | Anything you reuse or test | Short logic passed into `sorted()`, `filter()`, `map()` |

**Rule of thumb:** if a lambda is hard to read, turn it into a normal function.

# 15. Mini Project: A Sales Toolkit

Build a small set of functions that work together. Each one does one job.

1. `revenue(item)`: returns units times price for one product. (You already have this.)
2. `category_total(sales, category)`: returns the total revenue for one category.
3. `top_products(sales, n=3)`: returns a list of the names of the `n` products with the highest revenue.
   Use `sorted()` with a lambda.
4. `print_summary(sales)`: prints the total revenue, the revenue for each category, and the top 3 products.
   It should **call** your other functions, not repeat their code.


````python
# START HERE
def revenue(item):
    return item["units"] * item["price"]

def category_total(sales, category):
    total = 0
    for item in sales:
        if item["category"] == category:
            total += revenue(item)
    return total

def top_products(sales, n=3):
    by_revenue = sorted(sales, key=lambda item: revenue(item), reverse=True)
    names = []
    for item in by_revenue[:n]:
        names.append(item["product"])
    return names

def print_summary(sales):
    total = 0
    for item in sales:
        total = total + revenue(item)
    print(f"Total revenue: {total:,}")

    categories = set()
    for item in sales:
        categories.add(item["category"])
    for cat in sorted(categories):
        print(f"  {cat:<12} {category_total(sales, cat):>10,}")

    print("Top 3:", top_products(sales))

print_summary(sales)    
````
